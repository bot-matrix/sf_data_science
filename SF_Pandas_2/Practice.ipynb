```python
prac1 = pd.read_csv('https://raw.githubusercontent.com/justmarkham/pandas-videos/master/data/ufo.csv')
display(prac1)
```

```python
# Прочитайте данные, сделайте преобразование времени к формату datetime и выполните задания ниже.
prac1['Time'] = pd.to_datetime(prac1['Time'], dayfirst=False)
prac1
```

```python
# В каком году отмечается наибольшее количество случаев наблюдения НЛО в США?
years_sold = prac1['Time'].dt.year
print(years_sold)
print('Mode year:', years_sold.mode()[0])
```

```python
# Найдите средний интервал времени (в днях) между двумя последовательными случаями наблюдения НЛО в штате Невада (NV).

# добавляем столбец с датой и фильтруем данные по штату Невада (NV)
prac1['Date'] = prac1['Time'].dt.date
ufo_nv = prac1[prac1['State'] == 'NV']

# выделяем дату, считаем разницу и находим среднее
average_interval = ufo_nv['Time'].dt.date.diff().mean()

print(f"Средний интервал между наблюдениями в Неваде: {average_interval} дней")
```
